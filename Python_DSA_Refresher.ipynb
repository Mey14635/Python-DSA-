```python
days=int(input("Enter any number of days "))

def second_in_day(days):
  #is a day we have 24*60*60 seconds
  seconds=days*86400
  print(seconds)

second_in_day(days)
```

```python
length=int(input("Enter the side length "))
def perimeter(length):
  calculate_perimeter=length+length+length+length
  print ("The perimeter is " + str(calculate_perimeter))
def area(length):
  calculate_area=length*2
  print ("The perimeter is " + str(calculate_area))
perimeter(length)
area(length)
```

```python
arr=[]
print("Enter 5 numeric values  ")
for i in range (5):
  #so here we have to get input from users for the array
  values=int(input(f"Enter value {i+1}: "))
  #Then we populate the array
  arr.append(values)
total=sum(arr)
average=total/len(arr)
print (f"The average of all the numbers is : {average}")
```
