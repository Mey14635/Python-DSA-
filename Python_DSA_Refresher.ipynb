```python
length=int(input("Enter the side length"))
def perimeter(length):
  calculate_perimeter=length+length+length+length
  print ("The perimeter is " + str(calculate_perimeter))
def area(length):
  calculate_area=length*2
  print ("The perimeter is " + str(calculate_area))
perimeter(length)
area(length)
```
